```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set global figure size and plot style for matplotlib
plt.rcParams['figure.figsize'] = (14, 7)
plt.style.use('fivethirtyeight')

# Import display function to format outputs in IPython/Jupyter notebooks
from IPython.display import display
```

```python
# 1. Load the CSV file into a DataFrame
df = pd.read_csv('penguins_size.csv')

# 2. Display the first 5 rows (using the imported display function)
print("--- First 5 rows of the dataset ---")
display(df.head())
```

```python
# 3. Get the number of rows and columns (.shape returns a tuple: (rows, columns))
rows, columns = df.shape
print("\n--- Dimensions ---")
print(f"Rows: {rows}")
print(f"Columns: {columns}")
```

```python
# 4. Count missing values (nulls) per column
print("\n--- Count of null values per column ---")
print(df.isnull().sum())
```

```python
# 5. Calculate the maximum percentage of null values
# Compute the null percentage for each column and extract the highest value
null_percentages = (df.isnull().sum() / len(df)) * 100
print("\n--- Maximum percentage of nulls ---")
print(f"The maximum percentage of null values in a single column is: {null_percentages.max():.2f}%")
```
